```python
df = pd.read_csv("Customer-Churn.csv")

df.head()
```

```python
conn = sqlite3.connect(":memory:")
```

```python
df.to_sql(
    "telco_churn",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """

SELECT *
FROM telco_churn
LIMIT 5

"""

pd.read_sql(query, conn)
```

# **Calculate churn rate**

```python
query = """

SELECT
    Churn,
    COUNT(*) AS Customers
FROM telco_churn
GROUP BY Churn

"""

pd.read_sql(query, conn)
```

# **Churn by Contract**

```python
query = """

SELECT
    Contract,
    Churn,
    COUNT(*) AS Total

FROM telco_churn

GROUP BY Contract, Churn

ORDER BY Contract

"""

pd.read_sql(query, conn)
```

# **Average Monthly Charges**

```python
query = """

SELECT

Churn,

ROUND(
AVG(MonthlyCharges),2
) AS Avg_Monthly

FROM telco_churn

GROUP BY Churn

"""

pd.read_sql(query, conn)
```

# **Churn by Payment Method**

```python
query = """

SELECT

PaymentMethod,

COUNT(*) AS Customers,

SUM(

CASE

WHEN Churn='Yes'

THEN 1

ELSE 0

END

)

AS Churned

FROM telco_churn

GROUP BY PaymentMethod

ORDER BY Churned DESC

"""

pd.read_sql(query, conn)
```
